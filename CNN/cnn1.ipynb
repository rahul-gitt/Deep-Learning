```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings

warnings.filterwarnings('ignore')

from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Perceptron
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Conv2D, Flatten, MaxPooling2D, Dropout
from tensorflow.keras.utils import to_categorical
```

```python
df = pd.read_csv('mnist_train.csv')
df_test = pd.read_csv('mnist_test.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
# preprocess
X_train = df.drop('label', axis = 1).values
y_train = df['label'].values
X_test = df.drop('label', axis= 1).values
y_test = df['label'].values
```

```python
X_train = X_train.astype('float32')/ 255.0
X_test = X_test.astype("float32")/ 255.0
```

```python
X_train_img = X_train.reshape(-1, 28, 28)
X_test_img = X_test.reshape(-1, 28, 28)
```

```python
y_train_cat = to_categorical(y_train, 10)
y_test_cat = to_categorical(y_test, 10)
```

Perceptron

```python
perceptron = Sequential([
    Flatten(input_shape= (28, 28)),
    Dense(10, activation='softmax')
])
```

```python
perceptron.compile(optimizer= 'sgd', loss= 'categorical_crossentropy', metrics= ['accuracy'])
```

```python
history_percep = perceptron.fit(X_train_img, y_train_cat, epochs= 5,
batch_size= 32, validation_data= (X_train_img,y_test_cat), verbose = 1)
```

```python
acc_percep = perceptron.evaluate(X_test_img, y_test_cat, verbose= 0)[1]
```

```python
acc_percep
```

ANN

```python
ann = Sequential([
    Flatten(input_shape = (28, 28)),
    Dense(128, activation='relu'),
    Dense(64, activation= 'relu'),
    Dense(10, activation= 'softmax')
])
```

```python
ann.compile(optimizer= 'adam', loss= 'categorical_crossentropy',  metrics= ['accuracy'])
```

```python
history_ann = ann.fit(X_train_img, y_train_cat, epochs= 5, batch_size= 32,
validation_data = (X_test_img, y_test_cat), verbose= 1)
```

```python
acc_ann = ann.evaluate(X_test_img, y_test_cat, verbose= 0)[1]
```
